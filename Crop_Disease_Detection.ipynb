```python

import os
import tensorflow as tf

plant_base = "/content/plant_images/resize plant_leaves dataset(1545)"
disease_base = "/content/disease_images"

os.makedirs(disease_base, exist_ok=True)

# Create one folder for each plant-disease combination
for plant in os.listdir(plant_base):
    plant_path = os.path.join(plant_base, plant)

    if not os.path.isdir(plant_path):
        continue

    for disease in os.listdir(plant_path):
        disease_path = os.path.join(plant_path, disease)

        if not os.path.isdir(disease_path):
            continue

        class_name = plant.split("(")[0].strip() + " - " + disease.split("(")[0].strip()
        class_name = class_name.replace("/", "-")
        target = os.path.join(disease_base, class_name)
        os.makedirs(target, exist_ok=True)

        for file in os.listdir(disease_path):
            if file.lower().endswith((".jpg", ".jpeg", ".png")):
                source = os.path.join(disease_path, file)
                destination = os.path.join(target, file)

                if not os.path.exists(destination):
                    import shutil
                    shutil.copy2(source, destination)

print("Disease-level folders created!")
print("Number of classes:", len([
    d for d in os.listdir(disease_base)
    if os.path.isdir(os.path.join(disease_base, d))
]))
```

```python

import tensorflow as tf

dataset_path = "/content/disease_images"

train_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="training",
    seed=42,
    image_size=(128, 128),
    batch_size=32
)

val_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path,
    validation_split=0.2,
    subset="validation",
    seed=42,
    image_size=(128, 128),
    batch_size=32
)

class_names = train_data.class_names

print("\nNumber of disease classes:", len(class_names))
print("Training dataset loaded successfully!")
```

```python

import tensorflow as tf

model = tf.keras.Sequential([
    tf.keras.layers.Input(shape=(128, 128, 3)),
    tf.keras.layers.Rescaling(1./255),

    tf.keras.layers.Conv2D(32, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(64, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Conv2D(128, (3, 3), activation="relu"),
    tf.keras.layers.MaxPooling2D(),

    tf.keras.layers.Flatten(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.3),
    tf.keras.layers.Dense(len(class_names), activation="softmax")
])

model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

model.summary()
```

```python

history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=10
)

print("Training completed!")
```
